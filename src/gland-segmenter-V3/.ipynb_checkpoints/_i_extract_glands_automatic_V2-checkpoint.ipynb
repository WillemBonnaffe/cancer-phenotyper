```python
#####################################
## FUNCTIONS FOR MODEL APPLICATION ##
#####################################

## numpy_to_device
## Goal:
## Convert a list of numpy arrays to tensors suitable for model input on a specific device.
## Inputs:
## X (list of numpy arrays): List of images in numpy array format.
## Outputs:
## tensors (list of torch.Tensor): List of tensors formatted for model input.
def numpy_to_device(X, device='cpu'):
    tensors = []
    for x in X:        
        x = cv2.resize(x, (1024, 1024))
        x = torch.tensor(x)        
        x = x.swapaxes(2, 1).swapaxes(1, 0)  # (H, W, C) --> (C, H, W)
        x = x.float()
        x = x[None, :, :, :]  # (C, H, W) --> (B=1, C, H, W)
        x = x.to(device)
        tensors += [x]        
    return tensors

## device_to_numpy
## Goal:
## Convert a list of tensors from a device back to numpy arrays.
## Inputs:
## X (list of torch.Tensor): List of tensors on a device.
## img_size_x (int): Width of image.
## img_size_y (int): Height of image.
## Outputs:
## images (list of numpy arrays): List of images converted to numpy format.
def device_to_numpy(X, img_size_x, img_size_y):
    images = []
    for x in X:
        x = x.cpu().numpy()
        x = x.swapaxes(0, 1).swapaxes(1, 2)    
        x = x.astype('uint8')
        x = cv2.resize(x, (img_size_y, img_size_x))
        images += [x]        
    return images

#
###
```

```python
#####################################
## FUNCTIONS FOR OBJECT EXTRACTION ##
#####################################

## Imports
# from f_wsi_reader import overlay_shifted_copies
import numpy as np
import cv2

## polish
## Goal:
## Apply erosion to the image to polish the objects.
## Inputs:
## img (numpy array): Image to apply erosion on.
## Outputs:
## img (numpy array): Eroded image.
def polish(img):
    ## Erosion
    kernel = np.ones((5, 5), np.uint8)  # Adjust the kernel size if needed
    img = cv2.erode(img, kernel, iterations=1)
    img = img.astype(np.uint8)

    # ## Diffusion 
    # for k in range(3):
    #     img = overlay_shifted_copies(img) > 0
    # img = img.astype(np.uint8)
    # img = img.reshape(img.shape[0], img.shape[1], 1)
    
    return img

## extract_object_pixels
## Goal:
## Extract the pixel coordinates of objects from a binary segmentation mask.
## Inputs:
## mask (numpy.ndarray): A 2D binary segmentation mask where objects have a value of 1 (or 255) and background is 0.
## Outputs:
## object_pixels (list): A list of numpy arrays, each containing the (row, column) coordinates of the pixels of each object.
def extract_object_pixels(mask):
    ## Ensure the mask is binary (values are either 0 or 1)
    if mask.max() > 1:
        mask = mask // 255  # Convert 255 values to 1 for consistency if it's an 8-bit mask.

    ## Find connected components (objects) in the mask
    num_labels, labels = cv2.connectedComponents(mask.astype(np.uint8))

    ## Extract pixel coordinates for each object, ignoring the background (label 0)
    object_pixels = []
    for label in range(1, num_labels):
        ## Find pixel coordinates where the label is present
        pixels = np.column_stack(np.where(labels == label))
        object_pixels.append(pixels)

    return object_pixels

## get_bounding_box
## Goal:
## Get the bounding box of an object from its pixel coordinates.
## Inputs:
## coords (numpy.ndarray): Array of (row, column) coordinates of the object's pixels.
## Outputs:
## bounding_box (tuple): Bounding box coordinates (x_min, y_min, x_max, y_max).
def get_bounding_box(coords):
    y_min, x_min = coords.min(axis=0)
    y_max, x_max = coords.max(axis=0)
    return x_min, y_min, x_max, y_max

## get_bounding_box_square
## Goal:
## Get the bounding box of an object from its pixel coordinates, making it square.
## Inputs:
## coords (numpy.ndarray): Array of (row, column) coordinates of the object's pixels.
## Outputs:
## bounding_box (tuple): Bounding box coordinates (x_min, y_min, x_max, y_max).
def get_bounding_box_square(coords):
    y_min, x_min = coords.min(axis=0)
    y_max, x_max = coords.max(axis=0)
    
    ## Calculate width and height
    width = x_max - x_min
    height = y_max - y_min
    
    ## Make the bounding box square by taking the larger of the two dimensions
    side_length = max(width, height)
    
    ## Center the square box around the object's bounding box
    x_center = (x_min + x_max) // 2
    y_center = (y_min + y_max) // 2
    
    x_min = x_center - side_length // 2
    x_max = x_center + side_length // 2
    y_min = y_center - side_length // 2
    y_max = y_center + side_length // 2

    return x_min, y_min, x_max, y_max

## inlay_objects
## Goal:
## Visualise individual objects in an image.
## Inputs:
## img (numpy array): Image as a numpy array.
## coordinates (list): List of coordinates of pixels of objects.
## Outputs:
## img (numpy array): Image with objects inlaid in different colors.
def inlay_objects(img, coordinates):
    img = img * 0
    if len(img.shape) < 3:
        img = img[:, :, None]
    if img.shape[2] == 1: 
        img = np.repeat(img, 3, axis=2)
    for i in range(len(coordinates)):
        coordinates_i = coordinates[i].astype("int")
        color_vect = np.random.uniform(0, 1, 3) * 255
        img[coordinates_i[:, 0], coordinates_i[:, 1], 0] = color_vect[0]
        img[coordinates_i[:, 0], coordinates_i[:, 1], 1] = color_vect[1]
        img[coordinates_i[:, 0], coordinates_i[:, 1], 2] = color_vect[2]
    return img

#
###
```

```python
#############################################
## FUNCTIONS FOR GLAND EXTRACTION FROM WSI ##
#############################################

## load_model
## Goal:
## Load the machine learning model for segmentation.
## Inputs:
## model_path (str): Path to the model file.
## num_classes (int): Number of classes for segmentation.
## device (str): Computational device for model inference.
## Outputs:
## model (torch.nn.Module): Loaded model.
def load_model(model_path, num_classes, device):
    from f_model_transformer import UNetTransformer as Model
    model = Model(n_classes=num_classes).to(device)
    if os.path.exists(model_path):
        model.load_state_dict(torch.load(model_path))
        print("Loaded model")
    return model

## load_slide_and_get_tile_map
## Goal:
## Load slide and generate the tissue mask with the specified tile size.
## Inputs:
## wsi_path (str): Path to the whole slide image.
## tile_size_mask (int): Size of tiles used for mask generation.
## Outputs:
## slide: Loaded slide.
## mask: Generated mask for the slide.
## canvas_size: Dimensions of the canvas.
def load_slide_and_get_tile_map(wsi_path, tile_size_mask):
    slide = read_slide(wsi_path)
    slide_size = get_dimensions(slide)
    print(f'Slide size: {slide_size}')
    
    ## Get tile map and mask
    thumbnail, tile_map, canvas_size = get_tile_map(slide, tile_size_mask)
    mask = apply_all_filters(tile_map, low=10, std=15, high=200)
    mask = apply_all_transforms(mask)
    
    return slide, mask, canvas_size

## extract_patch_coordinates
## Goal:
## Extract coordinates of tiles containing tissue.
## Inputs:
## tile_map: The tissue map of the slide.
## tile_size_mask (int): Size of tiles used for mask generation.
## canvas_size (tuple): Dimensions of the canvas.
## Outputs:
## tile_coordinates (list): List of coordinates for tiles containing tissue.
def extract_patch_coordinates(tile_map, tile_size_mask, canvas_size):
    tile_coordinates = (np.argwhere(tile_map[:,:,0].T > 0) * tile_size_mask).astype(int)
    tile_coordinates = [
        [coord[0], coord[1]] 
        for coord in tile_coordinates
        if 0 < coord[0] < canvas_size[1] and 0 < coord[1] < canvas_size[0]
    ]
    return tile_coordinates

## process_patches_in_batches
## Goal:
## Process patches in batches, applying the model and collecting predictions.
## Inputs:
## tile_coordinates (list): List of coordinates for tiles containing tissue.
## wsi_path (str): Path to the whole slide image.
## tile_size_model (int): Size of tiles used for model input.
## batch_size (int): Number of patches to process in each batch.
## level (int): Downsampling level for patch extraction.
## model: The loaded machine learning model.
## device (str): Computational device for model inference.
## downsample_preds (bool): Whether to downsample predictions.
## num_classes (int): Number of classes for segmentation. Default is 3.
## Outputs:
## preds (list): List of model predictions for patches.
def process_patches_in_batches(tile_coordinates, wsi_path, tile_size_model, batch_size, level, model, device, downsample_preds, num_classes):
    preds = []
    ds = 1 if level == 0 else 4 * level
    tile_size_ = int(tile_size_model / ds)
    
    for i in range(0, len(tile_coordinates), batch_size):
        batch_coords = tile_coordinates[i:i + batch_size]
        patches = slide_path_to_tiles_at_coordinates(wsi_path, batch_coords, tile_size_, level)
        
        ## Convert patches to the correct device
        X = numpy_to_device(patches, device)
        
        with torch.no_grad():
            batch_preds = []
            for x in X:
                Yhat = model(x)
                Yhat = torch.sigmoid(Yhat) * 255
                Yhat = device_to_numpy(Yhat, tile_size_model, tile_size_model)[0]
                if num_classes > 1:
                    Yhat = np.argmax(Yhat, axis=2).reshape(Yhat.shape[0], Yhat.shape[1], 1)
                if downsample_preds:
                    Yhat = cv2.resize(Yhat, (Yhat.shape[1] // 2, Yhat.shape[0] // 2))
                batch_preds.append(Yhat)
        
        preds.extend(batch_preds)
    
    return preds

## reassemble_predictions
## Goal:
## Reassemble predictions into a full mask matching the whole slide dimensions.
## Inputs:
## preds (list): List of model predictions for patches.
## tile_coordinates (list): List of coordinates for tiles.
## canvas_size (tuple): Dimensions of the canvas.
## Outputs:
## mask: Reassembled mask.
def reassemble_predictions(preds, tile_coordinates, canvas_size):
    mask = reassemble_tiles(preds, tile_coordinates, canvas_size[1], canvas_size[0])
    mask = mask[:, :, 0].reshape(mask.shape[0], mask.shape[1], 1)
    return mask

## polish_and_extract_objects
## Goal:
## Polish the mask and extract object coordinates.
## Inputs:
## mask: Reassembled mask.
## tissue_mask: The tissue mask of the slide.
## ds (int): Downscale factor for mask polishing.
## num_polish (int): Number of times to apply mask polishing.
## Outputs:
## mask_bin: Polished binary mask.
## rescaled_gland_coords: Coordinates of extracted objects.
def polish_and_extract_objects(mask, tissue_mask, ds, num_polish):
    ## Downsample mask
    mask_bin = ((mask > 0) * 1).astype('uint8')
    mask_bin = cv2.resize(mask_bin, (mask_bin.shape[1] // ds, mask_bin.shape[0] // ds))
    mask_bin = mask_bin[:, :, None]
    
    ## Polish mask
    for _ in range(num_polish):
        mask_bin = polish(mask_bin)
    
    ## Extract objects
    gland_coords = extract_object_pixels(mask_bin)
    rescaled_gland_coords = [coords * ds for coords in gland_coords]
    
    return mask_bin, rescaled_gland_coords

## save_extracted_objects
## Goal:
## Save extracted objects as images based on the bounding box coordinates.
## Inputs:
## rescaled_coords: Coordinates of extracted objects.
## wsi_path (str): Path to the whole slide image.
## buffer (int): Minimum buffer distance for bounding box.
## dxy (int): Padding around extracted objects.
## output_folder (str): Path to save the extracted objects.
## tile_size_model (int): Size of tiles used for model input.
def save_extracted_objects(rescaled_coords, wsi_path, buffer, dxy, output_folder, tile_size_model):
    for i, coords in enumerate(rescaled_coords):
        ## Discard small objects
        if len(coords) > 1000:  
            x_min, y_min, x_max, y_max = get_bounding_box_square(coords)
            
            ## Reload the relevant patches to extract the object
            if x_min - buffer > 0 and x_max + buffer < tile_size_model and y_min - buffer > 0 and y_max + buffer < tile_size_model:
                object_img = reload_patch_for_object(wsi_path=wsi_path, x_min=x_min, y_min=y_min, x_max=x_max, y_max=y_max, dxy=dxy)
                object_filename = os.path.join(output_folder, f'{os.path.basename(wsi_path)}_object_{i}.png')
                cv2.imwrite(object_filename, object_img)

## reload_patch_for_object
## Goal:
## Reload relevant patches based on bounding box coordinates to extract object images.
## Inputs:
## wsi_path (str): Path to the whole slide image.
## x_min, y_min, x_max, y_max (int): Bounding box coordinates.
## dxy (int): Padding around bounding box.
## Outputs:
## object_img: Extracted object image.
def reload_patch_for_object(wsi_path, x_min, y_min, x_max, y_max, dxy):
    ## Placeholder for patch reloading logic based on coordinates
    object_img = extract_patch(wsi_path, x_min - dxy, y_min - dxy, x_max + dxy, y_max + dxy)
    object_img = cv2.cvtColor(object_img, cv2.COLOR_BGR2RGB)
    return object_img

## process_whole_slide_images
## Goal:
## Process a list of whole slide images, extract tissue regions, apply a machine learning model for segmentation,
## refine results, and save extracted objects.
## Inputs:
## wsi_path (list): Path to whole slide images.
## output_folder (str): Path to the folder where the extracted objects will be saved.
## tile_size_mask (int): Tile size for tissue segmentation and coordinate extraction. Default is 256.
## tile_size_model (int): Tile size for patch extraction for model input. Default is 1024.
## batch_size (int): Number of patches processed in each batch. Default is 8.
## model_path (str): Path to the pre-trained model file. Default is 'models/TransUNet_V2_0.pth'.
## device (str): Computational device used for model inference. Default is 'cuda'.
## num_classes (int): Number of classes for segmentation. Default is 3.
## buffer (int): Minimum buffer distance for bounding box around objects. Default is 200.
## dxy (int): Extra padding around extracted objects. Default is 30.
## level (int): Downsampling level for patch extraction. Default is 0.
## downsample_preds (bool): Whether to downsample the model predictions. Default is True.
## ds (int): Downscale factor for mask polishing. Default is 4.
## num_polish (int): Number of times to apply mask polishing. Default is 1.
## Outputs:
## None
def process_whole_slide_images(
    wsi_path,
    output_folder,
    tile_size_mask=256,
    tile_size_model=1024,
    batch_size=8,
    model_path='models/TransUNet_V2_0.pth',
    device='mps',
    num_classes=3,
    buffer=200,
    dxy=30,
    level=0,
    downsample_preds=True,
    ds=4,  # Downscale factor for mask polishing
    num_polish=1  # Number of times to apply mask polishing
):

    ## Create output folder if it does not exist
    os.makedirs(output_folder, exist_ok=True)

    ## Load the model
    model = load_model(model_path=model_path, num_classes=num_classes, device=device)


    ## Load slide and get tile map
    slide, tile_map, canvas_size = load_slide_and_get_tile_map(wsi_path=wsi_path, tile_size_mask=tile_size_mask)

    ## Extract patch coordinates
    tile_coordinates = extract_patch_coordinates(tile_map=tile_map, tile_size_mask=tile_size_mask, canvas_size=canvas_size)

    ## Process patches in batches and apply the model
    preds = process_patches_in_batches(
        tile_coordinates=tile_coordinates, 
        wsi_path=wsi_path, 
        tile_size_model=tile_size_model, 
        batch_size=batch_size, 
        level=level, 
        model=model, 
        device=device, 
        downsample_preds=downsample_preds,
        num_classes=num_classes
    )

    ## Reassemble predictions into a full mask
    mask = reassemble_predictions(preds=preds, tile_coordinates=tile_coordinates, canvas_size=canvas_size)

    ## Polish the mask and extract objects
    mask_bin, rescaled_gland_coords = polish_and_extract_objects(mask=mask, tissue_mask=tile_map, ds=ds, num_polish=num_polish)

    ## Save extracted objects
    save_extracted_objects(
        rescaled_coords=rescaled_gland_coords, 
        wsi_path=wsi_path, 
        buffer=buffer, 
        dxy=dxy, 
        output_folder=output_folder, 
        tile_size_model=tile_size_model
    )

    print('')
    print('Done')
```
