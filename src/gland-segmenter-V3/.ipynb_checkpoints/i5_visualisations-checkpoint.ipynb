```python
PTIF = '/Users/user/Documents/projects/GlandSeg/data/examples/'
FILENAME = 'TCGA-HC-7213-01Z-00-DX1.5b03a3f2-ae77-4906-81b6-e9a5c0c5ed16.svs__4256_4864_14624_15232.png'
img = cv2.imread(PTIF + f"in_tmp/{FILENAME}")
mask_1 = cv2.imread(PTIF + f"out_tmp_glands/{FILENAME}")
mask_2 = cv2.imread(PTIF + f"out_tmp_nuclei/{FILENAME}")

## Check size
if mask_1.shape != img.shape:
    mask_1 = cv2.resize(mask_1, (img.shape[1], img.shape[0]))    
if mask_2.shape != img.shape:
    mask_2 = cv2.resize(mask_2, (img.shape[1], img.shape[0]))

fig, axes = plt.subplots(2, 2, figsize=(12,12))
axes[0,0].imshow(img)
axes[0,1].imshow(mask_2)
axes[1,0].imshow(mask_1)
# axes[1,1].imshow((mask_1 * mask_2))
# axes[1,1].imshow(mask_1 * (mask_2 > 150).astype(np.uint8))
axes[1,1].imshow(mask_1 * (1 - (mask_2 > 150)).astype(np.uint8))
plt.show()
plt.close()
```
