# Lazy dataloader

```python
##############
## INITIATE ##
##############

## Imports
import numpy as np
import cv2
import matplotlib.pyplot as plt
import torch
from torch.utils.data import Dataset, DataLoader

## Import modules
from f_dataset import SegmentationDataset
from f_dataset import augment

#
###
```

```python
###############
## LOAD DATA ##
###############

## Parameters
path_to_dataset = '/Volumes/SED/BDI/databases/RAW/datasets/SEGG/images/train_tiles/'
# thresholds = np.array([0, 150, 210, 255]) # Example thresholds
thresholds = np.array([0, 150, 255]) # Example thresholds
batch_size = 1

## Create dataset and dataloader
dataset = SegmentationDataset(path_to_dataset, thresholds, transform=augment)
dataloader = DataLoader(dataset, batch_size=batch_size, shuffle=True, num_workers=4)

def mps_to_numpy(X):
    images = []
    for x in X:
        x = x.cpu().numpy()
        x = x.swapaxes(0, 1).swapaxes(1, 2)
        x = x.astype('uint8')        
        images += [x]        
    return images

## Example of iterating through the DataLoader
images = []
masks = []
k = 0
with torch.no_grad():
    for X, Y in dataloader:        

        ## Inputs
        images += [mps_to_numpy(X)[0]]
        
        ## Target
        # Y = Y[:, None, :, :]
        masks += [mps_to_numpy(Y)[0]]        
        
        k += 1
        if k > 16:
            break

#
###
```

```python
###########################
## VISUALISE PREDICTIONS ##
###########################

## Visualise image
fig, axes = plt.subplots(4, 4, figsize=(9,9))
k = 0
for i in range(4):
    for j in range(4):
        axis = axes[i,j]        
        axis.imshow(images[k])
        k += 1
plt.tight_layout()
plt.show()
plt.close()

## Visualise image
fig, axes = plt.subplots(4, 4, figsize=(9,9))
k = 0
for i in range(4):
    for j in range(4):
        axis = axes[i,j]
        axis.imshow(masks[k])
        k += 1
plt.tight_layout()
plt.show()
plt.close()

#
###
```

```python
## pad_image_even
## Goal:
## Pad image to an even size.
## Inputs: 
## img (numpy array): Image to pad.
## Outputs:
## img_padded (numpy array): Image padded.
def pad_image_even(img):    
    pad_size = np.max(img.shape)
    if (img.shape[0] < pad_size):
        img_padded = np.concatenate((img, np.zeros((pad_size - img.shape[0], img.shape[1], img.shape[2]))), axis=0)
    if (img.shape[1] < pad_size):
        img_padded = np.concatenate((img, np.zeros((img.shape[0], pad_size - img.shape[1], img.shape[2]))), axis=1)
    return img_padded.astype(int)

## Pad even
print(img.shape)
img_padded = pad_image_even(img)
img_padded = cv2.resize(img_padded.astype('uint8'), (1024, 1024), interpolation=cv2.INTER_LINEAR)
print(img_padded.shape)

## Visualise image
plt.imshow(img_padded)
plt.show()
plt.close()

## Pad even
print(mask.shape)
mask_padded = pad_image_even(mask)
mask_padded = cv2.resize(mask_padded.astype('uint8'), (1024, 1024), interpolation=cv2.INTER_LINEAR)
print(img_padded.shape)

## Visualise image
plt.imshow(mask_padded)
plt.show()
plt.close()
```

```python
## labelise_channels
## Goal: 
## Labelise channels of image according to a threshold array.
## Inputs:
## img (numpy array): Image to split the channels of.
## thresholds (numpy array): Thresholds according to which channels in the image should be split.
## Outputs:
## img_new (numpy array): Image with labelised channels.
def labelise_channels(img, thresholds):
    img_new = np.zeros((img.shape[0], img.shape[1], img.shape[2]))
    for i in range(len(thresholds) - 1):
        img_new[:, :, i] = ((img[:, :, i] >= thresholds[i]) & (img[:, :, i] < thresholds[i + 1])) * i
    img_new = np.sum(img_new, axis=2).reshape(img.shape[0], img.shape[1], 1) # Format
    return img_new

## Labelise channels
thresholds = np.array([0, 150, 210, 255])
mask_padded_labelised = labelise_channels(mask_padded, thresholds)

## Visualise image
plt.imshow(mask_padded_labelised)
plt.show()
plt.close()
```

```python
## Import module
from f_dataset_preload import cropflip_transform, power_transform, brightness_transform

## Convert to tensor
X = torch.tensor(img_padded)
X = X.swapaxes(2, 1).swapaxes(1, 0)
X = X.float()
print(X.shape)
Y = torch.tensor(mask_padded_labelised)
Y = Y.swapaxes(2, 1).swapaxes(1, 0)
Y = Y.float()
print(Y.shape)

## Transform parameters
alpha = np.random.uniform(0.2, 1.5)
flip_h = int(np.round(np.random.uniform(0, 1, 1))) == 1
flip_v = int(np.round(np.random.uniform(0, 1, 1))) == 1

## Apply transform
X = cropflip_transform(X, alpha=alpha, target_size=X.shape[1], flip_h=flip_h, flip_v=flip_v) 
Y = cropflip_transform(Y, alpha=alpha, target_size=Y.shape[1], flip_h=flip_h, flip_v=flip_v) 

## Colour transforms
X = power_transform(X, strength=0.05)
X = brightness_transform(X, strength=0.05)

## Convert back to numpy
X = X.numpy()
X = X.swapaxes(0, 1).swapaxes(1, 2)
Y = Y.numpy()
Y = Y.swapaxes(0, 1).swapaxes(1, 2)

## Convert type
X = X.astype("uint8")

## Visualise image
plt.imshow(X)
plt.show()
plt.close()

## Visualise image
plt.imshow(Y)
plt.show()
plt.close()
```
