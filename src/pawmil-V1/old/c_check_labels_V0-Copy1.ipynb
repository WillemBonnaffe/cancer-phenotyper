```python
################
## GET LABELS ##
################

# Define the paths to the input and output files
pti = "/Volumes/SED/BDI/databases/RAW/datasets/PPCG/metadata/metadata_handformatted_v20240814.csv"  # Replace with the actual path to your CSV file
pto = "/Users/willembonnaffe/Documents/Academia/Projects/PAW-MIL-2D-PPCG/outputs/PPCG_MQ_SmallMediumNozeroNew_256_MultiTask_Augmented_all/"  # Replace with the actual path to your output folder

# Load the CSV file
df = pd.read_csv(pti)
print(df.shape)

# Define the columns to extract
labels_columns = [
    "clinical_lethality_index",
    "clinical_gleason_index_1",
    "clinical_gleason_index_2",
    "LOH.2.127395672.140774443",
    "LOH.5.96283821.109868889",
    "LOH.6.83769908.118663890",
    "LOH.8.10662555.35203860",
    "GAIN.8.91110300.145138636",
    "LOH.10.82181072.90887151",
    "HD.10.87579816.106758234",
    # "LOH.13.29920635.39841482",
    # "LOH.13.39841482.55871837",
    # "LOH.13.78637423.103808366",
    # "LOH.16.49820164.70578817",
    # "LOH.17.7157.18051975",
    # "LOH.21.39871276.42935269"
]
path_column = 'local_path'
case_id_column = 'qc_case_id'
center_column = 'Centre'

# Extract the relevant columns
labels = df[labels_columns]
paths = df[[path_column]]
case_ids = df[[case_id_column]]
centers = df[[center_column]]

# Threshold the labels to binarize them
labels = (labels > 0.5).astype(int)

# Repeat each row n times to match the augmented dataset
n = 4
labels_repeated = pd.concat([labels] * n, ignore_index=True)
paths_repeated = pd.concat([paths] * n, ignore_index=True)
case_ids_repeated = pd.concat([case_ids] * n, ignore_index=True)
centers_repeated = pd.concat([centers] * n, ignore_index=True)

# Save the extracted and repeated columns to separate text files with tab separation
labels_repeated.to_csv(f"{pto}/tab_labels.txt", index=False, header=False, sep='\t')
paths_repeated.to_csv(f"{pto}/tab_paths.txt", index=False, header=False, sep='\t')
case_ids_repeated.to_csv(f"{pto}/tab_groups.txt", index=False, header=False, sep='\t')
centers_repeated.to_csv(f"{pto}/tab_groups_2.txt", index=False, header=False, sep='\t')

#
###
```
