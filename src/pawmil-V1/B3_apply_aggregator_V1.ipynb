```python
##############################
## EVALUATE MULTIPLE MODELS ##
##############################

## Path to main work directory
pwd = "/Volumes/Elements/BDI/projects/GSG/outputs/o3_mil/tcga-oct/model_resnet_tcgaoct_raw_FTL_3/"

## Path to embeddings
pt_embeddings_file = pwd + "encoder/obj_embeddings.pkl"

## Move to aggregator directory
pwd = pwd + "aggregator/singletask-EMT.GAIN/"
hp_star = np.genfromtxt(pwd + "tab_best_hyperparameters.csv", delimiter=",", dtype="str")

## For each label
for hp_star_ in hp_star[1:]: # Skip header

    ## Strip hypertrain metadata
    label, fold, hp = hp_star_    
    bs, ne, lr = str.split(hp, "_")    
    bs = int(float(bs))
    ne = int(float(ne))

    ## Get labels
    pt_labels_file = pwd + f"{label}/tab_labels.txt"        

    ## Path to output
    pt_output_folder = pwd + f"{label}/predictions/"
    if os.path.exists(pt_output_folder) == False:
        os.mkdir(pt_output_folder)
    
    ## For each model
    for val_fold_idx_ in [0, 1, 2, 3, 4]:
        for repeat_idx_ in [0, 1, 2]:
            
            ## Get model id and path to model file
            model_id = f"mod_agg_fold{val_fold_idx_}_repeat{repeat_idx_}_bs{bs}_ne{ne}_lr{lr}"    
            pt_model_file = pwd + f"{label}/hypertrain_fold_{fold}/models/{model_id}.pth"

            ## Output subfolder
            pt_output_folder_ = pt_output_folder + f"{label}_hpt_fold_{fold}_{model_id}/"
            if os.path.exists(pt_output_folder_) == False:
                os.mkdir(pt_output_folder_)

            ## Message 
            print(f"{label}_hpt_fold_{fold}_{model_id}")
            
            ## Apply aggregator
            main_apply_aggregator_single(
                pt_embeddings_file,
                pt_labels_file, 
                pt_model_file,
                pt_output_folder_,
                device = "mps"
                )

#
###
```
