```python
#############################################
## MERGE EMBEDDINGS FROM DIFFERENT FOLDERS ##
#############################################

import pickle
import numpy as np
import os

## Main path
pwd = "/Users/willembonnaffe/Documents/Academia/Projects/PAW-MIL-2D-PPCG/outputs/"

## Path to output
pto = pwd + "PPCG_MQ_SmallMediumNozeroNew_256_MultiTask_Augmented_all_unique/"

## Initialize lists for embeddings, coordinates, and tables
embeds = []
coords = []
paths = []
groups = []
groups_2 = []
labels = []

## Loop through the 4 unique folders
for i in range(4):
    
    ## Path to subfolder
    subfolder = f"PPCG_MQ_SmallMediumNozeroNew_256_MultiTask_Augmented_p{i+1}_unique/"
    
    ## Load embeddings and coordinates
    embeds_ = pickle.load(open(pwd + subfolder + "encoder/obj_embeddings.pkl", "rb"))
    coords_ = pickle.load(open(pwd + subfolder + "encoder/obj_coordinates.pkl", "rb"))
    
    ## Load table files
    paths_ = np.genfromtxt(pwd + subfolder + "tab_paths.txt", dtype="str")
    groups_ = np.genfromtxt(pwd + subfolder + "tab_groups.txt", dtype="str")
    groups_2_ = np.genfromtxt(pwd + subfolder + "tab_groups_2.txt", dtype="str")
    labels_ = np.genfromtxt(pwd + subfolder + "tab_labels.txt", dtype="str")
    
    ## Check the lengths of loaded data
    print(f"Subfolder {i+1}: {len(embeds_)} embeddings, {len(coords_)} coordinates")
    
    ## Collect embeddings, coordinates, and table data
    embeds += embeds_
    coords += coords_
    paths += paths_.tolist()
    groups += groups_.tolist()
    groups_2 += groups_2_.tolist()
    labels += labels_.tolist()

## Check the total lengths after merging
print(f"Total: {len(embeds)} embeddings, {len(coords)} coordinates")

## Create the output directory if it doesn't exist
os.makedirs(pto, exist_ok=True)
os.makedirs(pto + "encoder/", exist_ok=True)

## Save the merged embeddings and coordinates
pickle.dump(embeds, open(pto + "encoder/obj_embeddings.pkl", "wb"))
pickle.dump(coords, open(pto + "encoder/obj_coordinates.pkl", "wb"))

## Save the merged table files
np.savetxt(pto + "tab_paths.txt", paths, fmt="%s")
np.savetxt(pto + "tab_groups.txt", groups, fmt="%s")
np.savetxt(pto + "tab_groups_2.txt", groups_2, fmt="%s")
np.savetxt(pto + "tab_labels.txt", labels, fmt="%s")

#
###
```
