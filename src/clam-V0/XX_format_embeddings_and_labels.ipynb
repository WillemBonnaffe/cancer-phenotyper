```python
##############
## INITIATE ##
##############

import pickle
import torch
import numpy as np
import pandas as pd
import os

#
###
```

```python
#################################
## FORMAT SLIDE IDS AND LABELS ##
#################################

## Paths
PTF_LABELS = "/Volumes/Elements/BDI/projects/GSG/outputs/o3_mil/tcga-oct/model_resnet_tcgaoct_baseline/aggregator/singletask-EMT.GAIN/tab_labels.txt"
PTF_PATHS = "/Volumes/Elements/BDI/projects/GSG/outputs/o3_mil/tcga-oct/model_resnet_tcgaoct_baseline/aggregator/singletask-EMT.GAIN/tab_paths.txt"
PTO = "/Volumes/Elements/BDI/projects/GSG/outputs/o4_clam/"
os.makedirs(PTO, exist_ok=True)

## Load files
labels = np.genfromtxt(PTF_LABELS)
slide_paths = np.genfromtxt(PTF_PATHS, dtype=str)

# Extract slide IDs (remove path and extension)
slide_ids = [path.split("/")[-1].replace(".svs", "") for path in slide_paths]

## Checks
# print(slide_ids)
# print(labels)

# Create DataFrame
df = pd.DataFrame({
    "slide_id": slide_ids,
    "label": labels
})

# Save to CSV
df.to_csv(PTO + "clam_slide_labels.csv", index=False)

#
###
```

```python
#######################
## FORMAT EMBEDDINGS ##
#######################

# Paths
PT_EMBEDS = "/Volumes/Elements/BDI/projects/GSG/outputs/o3_mil/tcga-oct/model_resnet_tcgaoct_raw_FTL_3/encoder/obj_embeddings.pkl"
PTO = "/Volumes/Elements/BDI/projects/GSG/outputs/o4_clam/embeddings/"
os.makedirs(PTO, exist_ok=True)


# Load data
with open(PT_EMBEDS, "rb") as f:
    embeddings = pickle.load(f)  # List of np.array, one per slide

# Save each embedding as a .pt file
for slide_id, features in zip(slide_ids, embeddings):
    assert isinstance(features, np.ndarray), f"Expected np.ndarray, got {type(features)}"
    torch.save(torch.from_numpy(features), os.path.join(PTO, f"{slide_id}.pt"))

#
###
```
