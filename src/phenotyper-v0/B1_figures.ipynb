```python
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

## Figure parameters
plt.rcParams['figure.dpi'] = 300
figsize = (14,5)
custom_params = {'axes.spines.right': False, 
                 'axes.spines.top': False, 
                 'axes.facecolor':'whitesmoke'
                }
sns.set_theme(style="ticks", rc=custom_params)

# list your folders
out_dirs = [
    '/Volumes/Elements/BDI/projects/GSG/figures/tcga-ffpes-ftl0',
    '/Volumes/Elements/BDI/projects/GSG/figures/tcga-ffpes-ftl1',
    '/Volumes/Elements/BDI/projects/GSG/figures/icgc-c-ftl0',
    '/Volumes/Elements/BDI/projects/GSG/figures/icgc-c-ftl1',
    '/Volumes/Elements/BDI/projects/GSG/figures/tcga-oct-ftl0',
    '/Volumes/Elements/BDI/projects/GSG/figures/tcga-oct-ftl1',
]

plt.figure(figsize=(5, 3))
for d in out_dirs:
    # load saved elbow data
    idxs = np.load(os.path.join(d, 'hac_elbow_idxs.npy'))
    dist = np.load(os.path.join(d, 'hac_elbow_distances.npy'))
    label = os.path.basename(d)  # e.g. 'tcga-ffpes-ftl0'
    plt.plot(idxs, dist, marker='o', label=label)

# plt.title('Elbow Plot for HAC - All Experiments')
plt.xlabel('Number of Clusters')
plt.ylabel('Ratio')
plt.grid(False)
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.legend(loc='best', frameon=False)
plt.show()
```
