```python
######################################
## IMPORTS AND LIBRARY DEPENDENCIES ##
######################################

## Imports
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
import pickle

#
###
```

```python
#############################
## SCRIPT CONFIGURATIONS   ##
#############################

## Define paths
pti = '/Users/user/Documents/projects/GlandSeg/outputs/V2/o2_embeddings/tcga_ffpes/model_resnet_tcga_seg_FTL_100/'
clustered_embeddings_path = pti + 'image_identifiers_and_reference_clustered.csv'  ## Path to the CSV file with embeddings and cluster labels
embeddings_path = pti + 'embeddings.npy'  ## Path to the embeddings (used to train k-NN)

## k-NN parameters
num_neighbours = 3  ## Set the number of nearest neighbors for k-NN

## Output path for saving the trained k-NN classifier
scaler_path = 'models/model_knn_scaler_seg_FTL_100.pkl'
knn_model_path = 'models/model_knn_classifier_seg_FTL_100.pkl'

#
###
```

```python
################################
## LOAD EMBEDDINGS AND LABELS ##
################################

## Load the clustered embeddings data from CSV
clustered_data = pd.read_csv(clustered_embeddings_path)

## Load the corresponding high-dimensional embeddings from the .npy file
embeddings = np.load(embeddings_path, allow_pickle=True)

## Extract the cluster labels from the CSV file
cluster_labels = clustered_data['cluster_label'].values

#
###
```

```python
############################
## STANDARDISE EMBEDDINGS ##
############################

## Standardize embeddings
scaler = StandardScaler()
standardized_embeddings = scaler.fit_transform(embeddings)  # Fit and transform

## Save the StandardScaler
with open(scaler_path, 'wb') as f:
    pickle.dump(scaler, f)

print("StandardScaler saved successfully.")

#
###
```

```python
##################################
## TRAIN THE K-NN CLASSIFIER    ##
##################################

## Initialize the k-NN classifier
knn_classifier = KNeighborsClassifier(n_neighbors=num_neighbours)

## Train the k-NN classifier using the high-dimensional embeddings and cluster labels
knn_classifier.fit(standardized_embeddings, cluster_labels)

## Check training success
print(f'k-NN classifier trained on {len(embeddings)} samples with {num_neighbours} neighbors')

#
###
```

```python
###################################
## SAVE THE TRAINED K-NN MODEL   ##
###################################

## Save the trained k-NN classifier with pickle
with open(knn_model_path, 'wb') as f:
    pickle.dump(knn_classifier, f)

print('k-NN classifier saved successfully with pickle')

#
###
```
