```python
######################################
## IMPORTS AND LIBRARY DEPENDENCIES ##
######################################

## Imports
import os
import pickle
import numpy as np
import pandas as pd

## GLOBAL VARIABLES
FOLDER_NAME = 'TCGA-HC-A6AP-01Z-00-DX1.1E2C19B4-6757-488D-AFB4-71AE5FD6EC11.svs'
MODEL_NAME = 'model_resnet_tcga_seg_FTL_100'

## Define paths
scaler_path = 'models/model_knn_scaler_seg_FTL_100.pkl'
knn_model_path = 'models/model_knn_classifier_seg_FTL_100.pkl'
output_embeddings_path = f"/Users/user/Documents/projects/GlandSeg/outputs/V2/o2_embeddings/{FOLDER_NAME}/{MODEL_NAME}/embeddings.npy"
output_identifiers_path = f"/Users/user/Documents/projects/GlandSeg/outputs/V2/o2_embeddings/{FOLDER_NAME}/{MODEL_NAME}/image_identifiers.csv"
output_predictions_path = f"/Users/user/Documents/projects/GlandSeg/outputs/V2/o2_embeddings/{FOLDER_NAME}/{MODEL_NAME}/image_identifiers_and_predicted_clusters.csv"

#
###
```

```python
##############################################################
## LOAD SCALER & k-NN CLASSIFIER AND PREDICT CLUSTER LABELS ##
##############################################################

## Load the saved StandardScaler
with open(scaler_path, "rb") as f:
    scaler = pickle.load(f)

print("Loaded StandardScaler.")

## Load the saved k-NN classifier
with open(knn_model_path, "rb") as f:
    knn_classifier = pickle.load(f)

print("Loaded k-NN classifier.")

## Load the newly generated embeddings
new_embeddings = np.load(output_embeddings_path)  # Uses embeddings from this notebook

## Standardize the new embeddings using the loaded scaler
standardized_new_embeddings = scaler.transform(new_embeddings)

## Predict cluster labels using the k-NN classifier
predicted_clusters = knn_classifier.predict(standardized_new_embeddings)

## Load image identifiers
identifiers_df = pd.read_csv(output_identifiers_path)

## Add predicted cluster labels
identifiers_df["predicted_cluster"] = predicted_clusters

## Save to CSV
identifiers_df.to_csv(output_predictions_path, index=False)

print(f"Predicted cluster labels saved to {output_predictions_path}")

#
###
```
